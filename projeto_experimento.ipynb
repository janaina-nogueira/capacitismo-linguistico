```python

import re
import unicodedata
import pandas as pd

# Normalização: minúsculas e remoção de acentos
def normalizar_texto(texto):
    texto = unicodedata.normalize(
        "NFD", str(texto).lower()
    )
    return "".join(
        c for c in texto
        if unicodedata.category(c) != "Mn"
    )

# Reunir termos e expressões do dicionário original
dicionario = sorted(set(
    normalizar_texto(t.strip())
    for t in termos_capacitistas + expressoes_mwe
    if t.strip()
))

# Expressões completas, com limites de palavras
padrao = re.compile(
    r"(?<!\w)(?:" +
    "|".join(
        re.escape(t) for t in
        sorted(dicionario, key=len, reverse=True)
    ) +
    r")(?!\w)"
)

def identificar_termos(texto):
    texto = normalizar_texto(texto)
    return list(dict.fromkeys(
        m.group(0) for m in padrao.finditer(texto)
    ))

# Aplicar às 1.500 sentenças
df["termos_encontrados"] = df["text"].fillna("").apply(
    identificar_termos
)

df["presenca_termo_lexico"] = (
    df["termos_encontrados"].apply(
        lambda termos: int(len(termos) > 0)
    )
)

print("=== PRESENÇA DE TERMOS DO DICIONÁRIO ===")
print(df["presenca_termo_lexico"].value_counts())

print("\n=== PROPORÇÕES (%) ===")
print(
    df["presenca_termo_lexico"]
    .value_counts(normalize=True).mul(100).round(2)
)

print("\n=== EXEMPLOS ===")
display(
    df[[
        "id_anot",
        "text",
        "presenca_termo_lexico",
        "termos_encontrados",
        "capacitismo_majoritario_bin"
    ]].head(10)
)
```

```python

import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency, fisher_exact

# Selecionar variáveis
dados = df[
    ["presenca_termo_lexico", "capacitismo_majoritario_bin"]
].dropna().copy()

dados["presenca_termo_lexico"] = (
    dados["presenca_termo_lexico"].astype(int)
)
dados["capacitismo_majoritario_bin"] = (
    dados["capacitismo_majoritario_bin"].astype(int)
)

# Tabela de contingência
tabela = pd.crosstab(
    dados["presenca_termo_lexico"],
    dados["capacitismo_majoritario_bin"]
).reindex(index=[0, 1], columns=[0, 1], fill_value=0)

tabela.index = ["Sem termo", "Com termo"]
tabela.columns = ["Não capacitista", "Capacitista"]

print("=== TABELA DE CONTINGÊNCIA ===")
display(tabela)

print("\n=== PERCENTUAIS POR GRUPO ===")
display((tabela.div(tabela.sum(axis=1), axis=0)*100).round(2))

# Qui-quadrado sem correção de Yates
chi2, p_chi, gl, esperadas = chi2_contingency(
    tabela.values, correction=False
)

# Fisher se houver frequência esperada inferior a 5
if (esperadas < 5).any():
    odds_ratio, p = fisher_exact(tabela.values)
    teste = "Exato de Fisher"
else:
    p = p_chi
    teste = "Qui-quadrado de independência"

# V de Cramér
n = tabela.values.sum()
v_cramer = np.sqrt(
    chi2 / (n * min(tabela.shape[0]-1, tabela.shape[1]-1))
)

print("\n=== RESULTADOS ESTATÍSTICOS ===")
print(f"Teste utilizado: {teste}")
print(f"N: {n}")
print(f"Qui-quadrado: {chi2:.4f}")
print(f"Graus de liberdade: {gl}")
print(f"Valor-p: {p:.8f}")
print(f"V de Cramér: {v_cramer:.4f}")

print("\n=== INTERPRETAÇÃO ===")
if p < 0.05:
    print("Associação estatisticamente significativa.")
else:
    print("Associação não estatisticamente significativa.")

print("\nFrequências esperadas:")
display(pd.DataFrame(
    esperadas,
    index=tabela.index,
    columns=tabela.columns
).round(2))
```

Presença lexical × concordância

```python

from scipy.stats import mannwhitneyu

# Identificar os 13 anotadores
col_anotadores = sorted(
    [c for c in df.columns if c.startswith("resumo_anotador")],
    key=lambda x: int(x.replace("resumo_anotador", ""))
)

# Extrair a decisão binária de cada anotador
def extrair_binario(valor):
    if pd.isna(valor):
        return np.nan

    resposta = str(valor).strip().split(",")[0].strip().upper()

    if resposta == "SIM":
        return 1
    elif resposta in ["NÃO", "NAO"]:
        return 0
    return np.nan

votos = df[col_anotadores].apply(
    lambda coluna: coluna.map(extrair_binario)
)

# Verificar quantidade de votos válidos
df["n_votos_validos"] = votos.notna().sum(axis=1)

print("=== VOTOS VÁLIDOS POR SENTENÇA ===")
print(df["n_votos_validos"].value_counts().sort_index())

# Contagem de votos
df["votos_sim"] = votos.eq(1).sum(axis=1)
df["votos_nao"] = votos.eq(0).sum(axis=1)

# Concordância com a decisão predominante
df["concordancia_proporcao"] = (
    df[["votos_sim", "votos_nao"]].max(axis=1)
    / df["n_votos_validos"].replace(0, np.nan)
)

# Comparar com a classificação majoritária original
maioria_calculada = np.where(
    df["votos_sim"] > df["votos_nao"], 1,
    np.where(df["votos_nao"] > df["votos_sim"], 0, np.nan)
)

divergencias = (
    pd.Series(maioria_calculada, index=df.index)
    .ne(df["capacitismo_majoritario_bin"])
    & pd.Series(maioria_calculada, index=df.index).notna()
)

print("\nDivergências com a maioria original:",
      divergencias.sum())

# Separar os grupos
sem_termo = df.loc[
    df["presenca_termo_lexico"] == 0,
    "concordancia_proporcao"
].dropna()

com_termo = df.loc[
    df["presenca_termo_lexico"] == 1,
    "concordancia_proporcao"
].dropna()

# Estatísticas descritivas
resumo = pd.DataFrame({
    "Grupo": ["Sem termo", "Com termo"],
    "N": [len(sem_termo), len(com_termo)],
    "Média": [sem_termo.mean(), com_termo.mean()],
    "Mediana": [sem_termo.median(), com_termo.median()],
    "Desvio-padrão": [sem_termo.std(), com_termo.std()]
})

print("\n=== ESTATÍSTICAS DESCRITIVAS ===")
display(resumo.round(4))

# Mann-Whitney
u, p = mannwhitneyu(
    com_termo,
    sem_termo,
    alternative="two-sided",
    method="asymptotic"
)

# Correlação bisserial de postos
# Positivo: maior concordância no grupo COM termo
r_rb = 2 * u / (len(com_termo) * len(sem_termo)) - 1

print("\n=== TESTE MANN-WHITNEY ===")
print(f"U = {u:.4f}")
print(f"p = {p:.8f}")
print(f"r_rb = {r_rb:.4f}")

print("\n=== INTERPRETAÇÃO ===")
if p < 0.05:
    print("Diferença estatisticamente significativa.")
else:
    print("Diferença não estatisticamente significativa.")

if r_rb > 0:
    print("Tendência de maior concordância COM termo.")
elif r_rb < 0:
    print("Tendência de maior concordância SEM termo.")
else:
    print("Ausência de diferença na direção do efeito.")
```

Concordancia global

```python

from statsmodels.stats.inter_rater import fleiss_kappa

# Organizar os 13 anotadores
col_anotadores = sorted(
    [c for c in df.columns if c.startswith("resumo_anotador")],
    key=lambda x: int(x.replace("resumo_anotador", ""))
)

# Separar as três dimensões de cada anotação
binario = pd.DataFrame(index=df.index)
categoria = pd.DataFrame(index=df.index)
polaridade = pd.DataFrame(index=df.index)

for col in col_anotadores:
    partes = df[col].fillna("").astype(str).str.split(
        ",", n=2, expand=True
    ).reindex(columns=[0, 1, 2], fill_value="")

    binario[col] = partes[0].str.strip().str.upper()
    categoria[col] = partes[1].str.strip()
    polaridade[col] = partes[2].str.strip()

# Padronizar classificação binária
binario = binario.replace({"NAO": "NÃO"})

# Função para calcular Fleiss
def calcular_fleiss(dados, classes, nome):
    validos = dados.apply(
        lambda linha: linha.isin(classes).all(),
        axis=1
    )

    dados_validos = dados.loc[validos]

    if dados_validos.empty:
        print(f"{nome}: nenhuma sentença válida.")
        return None

    matriz = np.column_stack([
        dados_validos.eq(classe).sum(axis=1).to_numpy()
        for classe in classes
    ])

    kappa = fleiss_kappa(matriz, method="fleiss")

    print(f"\n=== {nome.upper()} ===")
    print(f"Sentenças utilizadas: {len(dados_validos)}")
    print(f"Categorias: {classes}")
    print(f"Kappa de Fleiss: {kappa:.4f}")

    return {
        "Dimensão": nome,
        "N": len(dados_validos),
        "Kappa de Fleiss": kappa
    }

resultados = []

# 1. Classificação binária
r = calcular_fleiss(
    binario,
    ["SIM", "NÃO"],
    "Classificação binária"
)
if r is not None:
    resultados.append(r)

# 2. Categoria (somente 13 respostas SIM)
r = calcular_fleiss(
    categoria,
    [
        "Insulto direto",
        "Uso metafórico",
        "Uso aparentemente neutro"
    ],
    "Categoria"
)
if r is not None:
    resultados.append(r)

# 3. Polaridade (somente 13 respostas SIM)
r = calcular_fleiss(
    polaridade,
    ["Negativo", "Neutro", "Positivo"],
    "Polaridade"
)
if r is not None:
    resultados.append(r)

print("\n=== RESUMO DOS RESULTADOS ===")
df_kappa = pd.DataFrame(resultados)
display(df_kappa.round(4))
```

Kappa de Cohen

```python

from itertools import combinations
from sklearn.metrics import cohen_kappa_score

# Utilizar as matrizes criadas na Célula 6:
# binario, categoria e polaridade

def calcular_cohen_pares(dados, classes, dimensao):
    resultados = []

    for a1, a2 in combinations(col_anotadores, 2):

        # Considerar somente respostas válidas dos dois anotadores
        validos = (
            dados[a1].isin(classes) &
            dados[a2].isin(classes)
        )

        pares = dados.loc[validos, [a1, a2]]

        if len(pares) < 2:
            continue

        kappa = cohen_kappa_score(
            pares[a1],
            pares[a2],
            labels=classes
        )

        resultados.append({
            "Dimensão": dimensao,
            "Anotador 1": a1,
            "Anotador 2": a2,
            "N": len(pares),
            "Kappa": kappa
        })

    return pd.DataFrame(resultados)


# 1. Classificação binária
cohen_binario = calcular_cohen_pares(
    binario,
    ["SIM", "NÃO"],
    "Classificação binária"
)

# 2. Categoria
cohen_categoria = calcular_cohen_pares(
    categoria,
    [
        "Insulto direto",
        "Uso metafórico",
        "Uso aparentemente neutro"
    ],
    "Categoria"
)

# 3. Polaridade
cohen_polaridade = calcular_cohen_pares(
    polaridade,
    ["Negativo", "Neutro", "Positivo"],
    "Polaridade"
)

# Consolidar os resultados
cohen_todos = pd.concat(
    [cohen_binario, cohen_categoria, cohen_polaridade],
    ignore_index=True
)

# Resumo por dimensão
resumo_cohen = (
    cohen_todos
    .groupby("Dimensão")
    .agg(
        Pares=("Kappa", "count"),
        Kappa_medio=("Kappa", "mean"),
        Mediana=("Kappa", "median"),
        Desvio_padrao=("Kappa", "std"),
        Kappa_minimo=("Kappa", "min"),
        Kappa_maximo=("Kappa", "max"),
        N_minimo=("N", "min"),
        N_maximo=("N", "max")
    )
    .reset_index()
)

print("=== RESUMO DO KAPPA DE COHEN ===")
display(resumo_cohen.round(4))

print("\n=== 10 PARES COM MENOR CONCORDÂNCIA BINÁRIA ===")
display(
    cohen_binario.nsmallest(10, "Kappa").round(4)
)

print("\n=== 10 PARES COM MAIOR CONCORDÂNCIA BINÁRIA ===")
display(
    cohen_binario.nlargest(10, "Kappa").round(4)
)
```

Teste Q de cochran

```python

from statsmodels.stats.contingency_tables import cochrans_q

# Utilizar a matriz binária criada anteriormente
# SIM = 1 | NÃO = 0

matriz_q = binario.replace({
    "SIM": 1,
    "NÃO": 0,
    "NAO": 0
}).astype(int)

# Teste Q de Cochran
resultado_q = cochrans_q(matriz_q.to_numpy())

print("=== TESTE Q DE COCHRAN ===")
print(f"Sentenças: {matriz_q.shape[0]}")
print(f"Anotadores: {matriz_q.shape[1]}")
print(f"Q = {resultado_q.statistic:.4f}")
print(f"Graus de liberdade = {matriz_q.shape[1] - 1}")
print(f"p = {resultado_q.pvalue:.10g}")

# Proporção de classificações positivas por anotador
resumo_anotadores = pd.DataFrame({
    "Anotador": matriz_q.columns,
    "SIM": matriz_q.sum(axis=0).values,
    "NÃO": (1 - matriz_q).sum(axis=0).values,
    "Percentual_SIM": (
        matriz_q.mean(axis=0).values * 100
    )
})

print("\n=== CLASSIFICAÇÕES POR ANOTADOR ===")
display(resumo_anotadores.round(2))

print("\n=== INTERPRETAÇÃO ===")
if resultado_q.pvalue < 0.05:
    print(
        "Existem diferenças estatisticamente significativas "
        "nas proporções de respostas SIM entre os anotadores."
    )
else:
    print(
        "Não foram identificadas diferenças estatisticamente "
        "significativas nas proporções de respostas SIM."
    )
```

```python

from itertools import combinations
from statsmodels.stats.contingency_tables import mcnemar
from statsmodels.stats.multitest import multipletests


resultados_mcnemar = []

for a1, a2 in combinations(matriz_q.columns, 2):
    tabela = pd.crosstab(
        matriz_q[a1],
        matriz_q[a2]
    ).reindex(index=[0, 1], columns=[0, 1], fill_value=0)

    b = tabela.loc[0, 1]
    c = tabela.loc[1, 0]

    # McNemar exato para poucos pares discordantes
    exato = (b + c) < 25

    resultado = mcnemar(
        tabela,
        exact=exato,
        correction=not exato
    )

    resultados_mcnemar.append({
        "Anotador 1": a1,
        "Anotador 2": a2,
        "Discordâncias": b + c,
        "p": resultado.pvalue
    })

df_mcnemar = pd.DataFrame(resultados_mcnemar)

# Correção de Holm
df_mcnemar["p_ajustado"] = multipletests(
    df_mcnemar["p"],
    method="holm"
)[1]

df_mcnemar["Significativo"] = (
    df_mcnemar["p_ajustado"] < 0.05
)

print("=== COMPARAÇÕES PAREADAS ===")
print("Total de pares:", len(df_mcnemar))
print(
    "Pares com diferenças significativas:",
    df_mcnemar["Significativo"].sum()
)

print("\n=== PARES COM MENORES VALORES-P AJUSTADOS ===")
display(
    df_mcnemar.sort_values("p_ajustado")
    .head(15)
    .round(6)
)
```

Consenso

```python


from scipy.stats import mannwhitneyu, chi2_contingency, fisher_exact

# Número de votos concordantes com a maioria
df["n_concordantes"] = df[
    ["votos_sim", "votos_nao"]
].max(axis=1)

# Proporção de concordância
df["concordancia_proporcao"] = (
    df["n_concordantes"] / 13
)

# Consenso forte: pelo menos 11 dos 13
df["consenso_forte"] = (
    df["n_concordantes"] >= 11
).astype(int)

# Unanimidade: 13 dos 13
df["unanimidade"] = (
    df["n_concordantes"] == 13
).astype(int)

# Resumo por classificação majoritária
resumo = df.groupby(
    "capacitismo_majoritario_bin"
).agg(
    N=("n_concordantes", "size"),
    Concordancia_media=("concordancia_proporcao", "mean"),
    Concordancia_mediana=("concordancia_proporcao", "median"),
    DP=("concordancia_proporcao", "std"),
    Consenso_forte=("consenso_forte", "sum"),
    Unanimidade=("unanimidade", "sum")
)

print("=== CONCORDÂNCIA POR CLASSE ===")
display(resumo.round(4))

# Mann-Whitney
cap = df.loc[
    df["capacitismo_majoritario_bin"] == 1,
    "concordancia_proporcao"
]

nao_cap = df.loc[
    df["capacitismo_majoritario_bin"] == 0,
    "concordancia_proporcao"
]

u, p = mannwhitneyu(
    cap, nao_cap,
    alternative="two-sided"
)

r_rb = 2*u/(len(cap)*len(nao_cap)) - 1

print("\n=== MANN-WHITNEY ===")
print(f"U = {u:.4f}")
print(f"p = {p:.10g}")
print(f"r_rb = {r_rb:.4f}")

# Associação com consenso forte e unanimidade
for variavel in ["consenso_forte", "unanimidade"]:

    tabela = pd.crosstab(
        df["capacitismo_majoritario_bin"],
        df[variavel]
    ).reindex(index=[0, 1], columns=[0, 1], fill_value=0)

    chi2, p_chi, gl, esperadas = chi2_contingency(
        tabela, correction=False
    )

    if (esperadas < 5).any():
        _, p_teste = fisher_exact(tabela.values)
        teste = "Fisher"
    else:
        p_teste = p_chi
        teste = "Qui-quadrado"

    v = np.sqrt(chi2 / tabela.values.sum())

    print(f"\n=== {variavel.upper()} ===")
    display(tabela)
    print(f"Teste: {teste}")
    print(f"Qui-quadrado: {chi2:.4f}")
    print(f"p = {p_teste:.10g}")
    print(f"V de Cramér = {v:.4f}")
```

Análise de Robustez (Leave-One-Rater-Out)

```python
# Matriz binária criada nas células anteriores
# 1 = SIM | 0 = NÃO
votos = matriz_q.copy()

# Classificação original com 13 votos
maioria_original = (
    votos.sum(axis=1) >= 7
).astype(int)

resultados_loo = []
decisoes_loo = pd.DataFrame(index=votos.index)

for anotador in votos.columns:

    # Remover individualmente o anotador
    votos_restantes = votos.drop(columns=anotador)

    # Quantidade de votos SIM entre os 12 restantes
    soma_sim = votos_restantes.sum(axis=1)

    # Nova classificação
    # 1 = SIM, 0 = NÃO, NaN = empate (6 x 6)
    nova_decisao = pd.Series(
        np.where(
            soma_sim > 6, 1,
            np.where(soma_sim < 6, 0, np.nan)
        ),
        index=votos.index
    )

    decisoes_loo[anotador] = nova_decisao

    preservadas = (
        nova_decisao == maioria_original
    ).sum()

    inversoes = (
        nova_decisao.notna() &
        (nova_decisao != maioria_original)
    ).sum()

    empates = nova_decisao.isna().sum()

    resultados_loo.append({
        "Anotador removido": anotador,
        "Preservadas": preservadas,
        "Inversões": inversoes,
        "Empates": empates,
        "Estabilidade (%)": preservadas / len(votos) * 100
    })

# Resumo por anotador
df_loo = pd.DataFrame(resultados_loo)

print("=== LEAVE-ONE-RATER-OUT ===")
display(df_loo.round(2))

# Estatísticas gerais
print("\n=== RESUMO GLOBAL ===")
print(f"Total de sentenças: {len(votos)}")
print(f"Anotadores avaliados: {len(votos.columns)}")

print(
    "Estabilidade média (%):",
    round(df_loo["Estabilidade (%)"].mean(), 2)
)

print(
    "Total de inversões:",
    df_loo["Inversões"].sum()
)

print(
    "Total de empates:",
    df_loo["Empates"].sum()
)

# Identificar sentenças afetadas
afetadas = (
    decisoes_loo.isna() |
    decisoes_loo.ne(maioria_original, axis=0)
).any(axis=1)

print("\n=== SENTENÇAS AFETADAS ===")
print("Sentenças afetadas:", afetadas.sum())
print(
    "Percentual:",
    round(afetadas.mean() * 100, 2)
)

# Distribuição de votos das sentenças afetadas
print("\n=== DISTRIBUIÇÃO DE VOTOS ===")
display(
    pd.DataFrame({
        "Votos SIM": votos.sum(axis=1),
        "Afetada": afetadas
    }).groupby(["Votos SIM", "Afetada"])
      .size()
      .unstack(fill_value=0)
)
```

Bootstrap

```python

import numpy as np
from statsmodels.stats.inter_rater import fleiss_kappa

rng = np.random.default_rng(42)
B = 10000

votos_array = matriz_q.to_numpy(dtype=int)
n_sentencas, n_anotadores = votos_array.shape

kappas_boot = np.empty(B)

for b in range(B):
    idx = rng.integers(
        0, n_sentencas, size=n_sentencas
    )

    votos_b = votos_array[idx]
    sim = votos_b.sum(axis=1)

    matriz_fleiss = np.column_stack([
        n_anotadores - sim,
        sim
    ])

    kappas_boot[b] = fleiss_kappa(
        matriz_fleiss,
        method="fleiss"
    )

# Atualizar somente os resultados do Kappa
df_boot["Kappa_Fleiss"] = kappas_boot

resumo_boot = pd.DataFrame({
    "Estimativa_bootstrap": df_boot.mean(),
    "IC95_inferior": df_boot.quantile(0.025),
    "IC95_superior": df_boot.quantile(0.975),
    "Desvio_padrao": df_boot.std()
})

print("=== BOOTSTRAP ===")
display(resumo_boot.round(4))
```

```python
import statsmodels.api as sm

from scipy.stats import chi2
from sklearn.metrics import roc_auc_score

# Preparação dos dados
dados_m1 = df[
    ["capacitismo_majoritario_bin", "presenca_termo_lexico"]
].dropna().copy()

dados_m1 = dados_m1.astype(int)

# Variável dependente
y = dados_m1["capacitismo_majoritario_bin"]

# Variável explicativa e intercepto
X = sm.add_constant(
    dados_m1[["presenca_termo_lexico"]]
)

# Ajuste do modelo
modelo_m1 = sm.Logit(y, X).fit(disp=False)

print("=== REGRESSÃO LOGÍSTICA M1 ===")
print(modelo_m1.summary())

# Odds ratios e IC 95%
ic = modelo_m1.conf_int()

tabela_or = pd.DataFrame({
    "Coeficiente": modelo_m1.params,
    "OR": np.exp(modelo_m1.params),
    "IC95_inferior": np.exp(ic[0]),
    "IC95_superior": np.exp(ic[1]),
    "p": modelo_m1.pvalues
})

print("\n=== ODDS RATIOS ===")
display(tabela_or.round(4))

# Pseudo-R² de McFadden
pseudo_r2 = modelo_m1.prsquared

# Teste global de razão de verossimilhanças
lr = modelo_m1.llr
p_lr = modelo_m1.llr_pvalue

# Probabilidades estimadas
probabilidades_m1 = modelo_m1.predict(X)

# AUROC (avaliação na própria amostra)
auroc_m1 = roc_auc_score(
    y, probabilidades_m1
)

print("\n=== AVALIAÇÃO DO MODELO M1 ===")
print(f"N = {len(dados_m1)}")
print(f"Pseudo-R² McFadden = {pseudo_r2:.4f}")
print(f"Razão de verossimilhanças = {lr:.4f}")
print(f"p do teste global = {p_lr:.10g}")
print(f"AUROC (amostra de ajuste) = {auroc_m1:.4f}")

# Probabilidade estimada para cada condição
print("\n=== PROBABILIDADES ESTIMADAS ===")

cenarios = pd.DataFrame({
    "const": [1, 1],
    "presenca_termo_lexico": [0, 1]
})

cenarios["Probabilidade_estimada"] = (
    modelo_m1.predict(cenarios)
)

cenarios["Situação"] = [
    "Sem termo do dicionário",
    "Com termo do dicionário"
]

display(
    cenarios[
        ["Situação", "Probabilidade_estimada"]
    ].round(4)
)
```
